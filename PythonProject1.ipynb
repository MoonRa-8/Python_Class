```python
#Visualization 1: What are the top complaint types? Residential Noise
Top_Complaint_Types = df['Complaint Type'].value_counts().head(10).sort_values(ascending=False)
Top_Complaint_Types.plot(kind='bar', color=['green'])
plt.title('Top 10 Complaint Types')
plt.xlabel('Complaint Type')
plt.ylabel('Number of Complaints')
plt.xticks(rotation=90)
plt.show()
```

```python
#Visualization 2: Which borough has the most complaints? Brooklyn
Borough_Complaints = df['Borough'].value_counts().sort_values(ascending=False)
plt.figure(figsize=(8, 5))
Borough_Complaints.plot(kind='bar', color=['blue'])
plt.title('Borough with Highest Complaints')
plt.ylabel('Number of Complaints')
plt.xlabel('Borough')
plt.xticks(rotation=45)
plt.show()
```

```python
df[df['Complaint Type'].str.contains('Noise', case=False, na=False)]
```

```python
Boroughs = ['BRONX', 'BROOKLYN', 'MANHATTAN', 'QUEENS', 'STATEN ISLAND']
Noise_Complaints = df[df['Complaint Type'].str.contains('Noise', case=False, na=False) & df['Borough'].isin(Boroughs)]
Noise_Complaints_by_Borough = Noise_Complaints['Borough'].value_counts()
plt.figure(figsize=(8, 5))
Noise_Complaints_by_Borough.plot(kind='bar', color=['red'])
plt.title('Noise Complaints by Borough')
plt.xlabel('Borough')
plt.ylabel('Number of Noise Complaints')
plt.xticks(rotation=45)
plt.show()
```

```python
Boroughs = ['BRONX', 'BROOKLYN', 'MANHATTAN', 'QUEENS', 'STATEN ISLAND']
Noise_Complaints = df[df['Complaint Type'].str.contains('Noise', case=False, na=False) & df['Borough'].isin(Boroughs)]
Noise_Complaints_by_Borough = Noise_Complaints['Borough'].value_counts()
print(Noise_Complaints_by_Borough)
```

```python
#One more piece of information: May 6, 2019 had the most complaints.
df['Created Date'] = pd.to_datetime(df['Created Date'])
Daily_Complaints = df['Created Date'].dt.date.value_counts().sort_values(ascending=False)
print(Daily_Complaints)
```
