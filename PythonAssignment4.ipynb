```python
#Data Visualization: Salary Distribution
df['salary'].plot(kind='hist', bins=20)
plt.title('Distribution of Salary')
plt.ylabel('Number of Employees')
plt.xlabel('Salary')
plt.show()
```

```python
#Data Visualization: Average Salary by Rank
df.groupby(['rank'])['salary'].mean().plot(kind='bar', color=['blue'])
plt.title('Average Salary by Rank')
plt.ylabel('Average Salary')
plt.xlabel('Rank')
plt.xticks(rotation=45)
plt.show()
```
